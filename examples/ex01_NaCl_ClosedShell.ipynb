```python
from pyEDA import * 
```

```python
import numpy as np
np.set_printoptions(formatter={"float": "{: 0.3f}".format})
```

```python
mol = gto.Mole(atom=[['Na', 0, 0, 0], ['Cl', 0, 0, 2.36]], basis='def2-svp').build()
```

```python
fragments_parameters = [[[0], 1], [[1], -1]]
fragments_list = []
for fr in fragments_parameters:
    frag = mol.copy()
    frag.atom = make_frag_geom(mol, fr[0])
    frag.unit = 'B'
    frag.charge = fr[1]
    frag.build()
    fragments_list.append(frag)
```

```python
method = 'b3lyp'
mf = scf.RKS(mol).run(xc=method)
mf_fragments_list = []
for fr in fragments_list:
    mf_fr = scf.RKS(fr).run(xc=method)
    mf_fragments_list.append(mf_fr)
```

```python
eda = ETS_NOCV(mf, mf_fragments_list)
eda.run()
```
