```python
import sys, os
sys.path.append(os.path.join(os.getcwd(), '..'))

from pyEDA_NOCV import * 
```

```python
mol = gto.Mole(atom=[['C', 0, 0, 0], ['O', 0, 0, 1.13]], basis='def2-svp').build()
```

```python
fragments_parameters = [[[0], 0, 2], [[1], 0, -2]]
fragments_list = []
for fr in fragments_parameters:
    frag = mol.copy()
    frag.atom = make_frag_geom(mol, fr[0])
    frag.unit = 'B'
    frag.charge = fr[1]
    frag.charge = fr[2]
    frag.build()
    fragments_list.append(frag)
```

```python
method = 'pbe'
mf = scf.UKS(mol)
mf.grids.level = 8
mf.run(xc=method)
mf_fragments_list = []
for fr in fragments_list:
    mf_fr = scf.UKS(fr)
    mf_fr.grids.level = 8
    mf_fr.run(xc=method)
    mf_fragments_list.append(mf_fr)
```

```python
eda = U_ETS_NOCV(mf, mf_fragments_list)
eda.run()
```
